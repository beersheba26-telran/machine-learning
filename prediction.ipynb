```python
import pandas as pd
import numpy as np
df = pd.read_csv('vehicles-cleaned.csv')
df.head(4)
```

```python
year_aggr = df.groupby("year",as_index=False)["price"].median().sort_values("price")
year_aggr = year_aggr[( year_aggr["price"] >= 2500) & (year_aggr["price"] <= 10000)]
X = year_aggr[["year"]]
y = year_aggr["price"]
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X, y)
```

```python
import numpy as np
predicted_price = model.predict(np.array([[2028]]))
predicted_price
```
