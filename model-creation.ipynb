```python
import pandas as pd
df = pd.read_csv("vehicles-cleaned.csv")
```

```python
year_aggr = df.groupby("year",as_index=False)["price"].median().sort_values("price")
year_aggr = year_aggr[( year_aggr["price"] >= 2500) & (year_aggr["price"] <= 10000)]
X = year_aggr[["year"]]
y = year_aggr["price"]
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X, y)
import joblib
joblib.dump(model, 'year-price-model.pkl')
```
