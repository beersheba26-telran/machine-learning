```python
import pandas as pd
df = pd.read_csv('vehicles-cleaned.csv')
df.head(4)
```

```python
# One pass over df - groups by manufacturer and calculates min, max, and count
stats = df.groupby('manufacturer')['price'].agg(['min', 'max', 'count'])
stats.head(4)
```

```python
 #Find the manufacturer with the most sales
most_sold_manufacturer = stats['count'].idxmax()
min_price = stats.loc[(most_sold_manufacturer, 'min')]
max_price = stats.loc[(most_sold_manufacturer, 'max')]
count = stats.loc[(most_sold_manufacturer, 'count')]
```

```python
#print the results
print(f"The most sold manufacturer is: {most_sold_manufacturer}")
print(f"Minimum price: {min_price}")
print(f"Maximum price: {max_price}")
print(f"Total sales: {count}")
```
