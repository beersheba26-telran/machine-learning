```python
import pandas as pd
required_cols = ["manufacturer", "model", "year", "type","odometer","state", "price"]
input_file = "vehicles.csv"
output_file = "vehicles-cleaned.csv"
first_chunk = True
for chunk in pd.read_csv(input_file, chunksize=100000, usecols=required_cols):
    chunk = chunk.dropna(subset=required_cols) # taking only rows with values in the required colons
    chunk = chunk[(chunk["price"] > 200) & (chunk["price"] < 200000) & (chunk["year"] > 1990) &
        (chunk["year"] <= 2026) & (chunk["odometer"] >= 0) & (chunk["odometer"] < 700000)]
    chunk.to_csv(output_file, mode="a", header=first_chunk, index=False)
    first_chunk = False
    
```
